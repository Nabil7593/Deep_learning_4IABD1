```python
import numpy as np
from scipy.stats import ttest_ind


mlp_accuracies = [0.60, 0.63, 0.54, 0.52, 0.52, 0.61, 0.56, 0.61]  
transformer_accuracies = [0.45, 0.55] 

# Test statistique de Welch
t_stat, p_value = ttest_ind(mlp_accuracies, transformer_accuracies, equal_var=False)

print(f"t-statistic: {t_stat}")
print(f"p-value: {p_value}")

# Interprétation
alpha = 0.05
if p_value < alpha:
    print("Il y a une différence statistiquement significative entre les modèles.")
else:
    print("Aucune différence significative entre les modèles.")
```
